```python
#Câu 4. Loại bỏ stopword khỏi văn bản
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

text = "Natural language processing is one of the most exciting fields of artificial intelligence."

words = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered = [w for w in words if w.lower() not in stop_words]

print(filtered)
```

```python
#Câu 5. In các từ không phải stopword
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

word_list = [
    "I","am","learning","Natural","Language",
    "Processing","with","Python"
]

result = []

for word in word_list:
    if word.lower() not in stop_words:
        result.append(word)

print(result)
```

```python
#Câu 6. Tìm định nghĩa và ví dụ của một từ bằng WordNet và Wikipedia
!pip install wikipedia
from nltk.corpus import wordnet
import wikipedia

word = input("Nhập từ: ")

print("\nĐịnh nghĩa trong WordNet:\n")

for syn in wordnet.synsets(word):
    print("Definition:", syn.definition())

    if syn.examples():
        print("Example:", syn.examples())

print("\nThông tin Wikipedia:\n")

try:
    print(wikipedia.summary(word, sentences=3))
except:
    print("Không tìm thấy.")
```

```python
#Câu 7. Tìm từ đồng nghĩa và trái nghĩa
from nltk.corpus import wordnet

word = input("Nhập từ: ")

synonyms = set()
antonyms = set()

for syn in wordnet.synsets(word):
    for lemma in syn.lemmas():
        synonyms.add(lemma.name())

        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())

print("Đồng nghĩa:")
print(synonyms)

print("\nTrái nghĩa:")
print(antonyms)
```

```python
#Câu 9. So sánh độ giống nhau của hai danh từ
from nltk.corpus import wordnet

word1 = "car"
word2 = "automobile"

syn1 = wordnet.synsets(word1, pos=wordnet.NOUN)[0]
syn2 = wordnet.synsets(word2, pos=wordnet.NOUN)[0]

print("Similarity:", syn1.wup_similarity(syn2))
```

```python
#Câu 10. So sánh độ giống nhau của hai động từ
from nltk.corpus import wordnet

v1 = wordnet.synsets("run", pos=wordnet.VERB)[0]

v2 = wordnet.synsets("walk", pos=wordnet.VERB)[0]

print(v1.wup_similarity(v2))
```

```python
#Câu 12. In 15 kết hợp ngẫu nhiên đầu tiên
from nltk.corpus import names
import random

labeled_names = (
    [(name, 'male') for name in names.words('male.txt')] +
    [(name, 'female') for name in names.words('female.txt')]
)

random.shuffle(labeled_names)

for item in labeled_names[:15]:
    print(item)
```

```python
#Câu 13. Trích xuất ký tự cuối cùng của mỗi tên
from nltk.corpus import names

labeled_names = (
    [(name, 'male') for name in names.words('male.txt')] +
    [(name, 'female') for name in names.words('female.txt')]
)

features = []

for (name, gender) in labeled_names:
    features.append((name[-1], gender))

print(features[:20])
```
